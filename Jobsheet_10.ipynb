# 1. SVM Classification (Dataset: diabetes.csv)

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load dataset asli sesuai arahan modul
df_diabetes = pd.read_csv('diabetes.csv')

# Cara penulisan pemisahan fitur yang lebih rapi
X_svm = df_diabetes.drop('Outcome', axis=1)
y_svm = df_diabetes['Outcome']

# 2. Split Data dengan test_size 30%
X_train_svm, X_test_svm, y_train_svm, y_test_svm = train_test_split(X_svm, y_svm, test_size=0.3, random_state=42)

# 3. Standarisasi Data
sc_svm = StandardScaler()
X_train_svm_scaled = sc_svm.fit_transform(X_train_svm)
X_test_svm_scaled = sc_svm.transform(X_test_svm)

# 4. Hyperparameter Tuning dengan kombinasi yang berbeda
param_grid_svm = {
    'C': [0.5, 1, 5, 10, 20],
    'gamma': ['scale', 'auto', 0.01, 0.05],
    'kernel': ['rbf', 'linear', 'sigmoid']
}

print("--- Memulai Grid Search SVM ---")
# Menambahkan n_jobs=-1 agar memproses menggunakan semua core CPU
grid_search_svm = GridSearchCV(SVC(random_state=42), param_grid_svm, cv=5, scoring='accuracy', n_jobs=-1)
grid_search_svm.fit(X_train_svm_scaled, y_train_svm)

# 5. Evaluasi Hasil
best_svm = grid_search_svm.best_estimator_
print(f"Kombinasi Parameter Terbaik SVM: {grid_search_svm.best_params_}")
print(f"Akurasi Terbaik SVM: {accuracy_score(y_test_svm, best_svm.predict(X_test_svm_scaled)) * 100:.2f}%\n")
```

# 2. SVR Regression (Dataset: Salary_Data.csv)

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVR
from sklearn.metrics import r2_score

# 1. Load dataset asli
df_salary = pd.read_csv('Salary_Data.csv')
X_svr = df_salary[['YearsExperience']].values
y_svr = df_salary['Salary'].values

# 2. Split Data
X_train_svr, X_test_svr, y_train_svr, y_test_svr = train_test_split(X_svr, y_svr, test_size=0.25, random_state=123)

# 3. Hyperparameter Tuning (Menambahkan parameter epsilon agar lebih advance)
param_grid_svr = {
    'C': [1, 10, 100, 1000, 5000],
    'gamma': ['scale', 'auto', 0.1, 0.05],
    'epsilon': [0.1, 0.5, 1.0],
    'kernel': ['rbf', 'linear']
}

print("--- Memulai Grid Search SVR ---")
grid_search_svr = GridSearchCV(SVR(), param_grid_svr, cv=5, scoring='r2', n_jobs=-1)
grid_search_svr.fit(X_train_svr, y_train_svr)

# 4. Evaluasi Hasil
best_svr = grid_search_svr.best_estimator_
print(f"Kombinasi Parameter Terbaik SVR: {grid_search_svr.best_params_}")
print(f"R-squared (R2) SVR Terbaik: {r2_score(y_test_svr, best_svr.predict(X_test_svr)):.4f}\n")
```

#3. Naïve Bayes (Dataset: iklan_sosmed.csv)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import MinMaxScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Load dataset asli dengan delimiter yang benar
df_sosmed = pd.read_csv('iklan_sosmed.csv', delimiter=';')
X_nb = df_sosmed.iloc[:, 2:-1].values
y_nb = df_sosmed.iloc[:, -1].values

# 2. Split Data
X_train_nb, X_test_nb, y_train_nb, y_test_nb = train_test_split(X_nb, y_nb, test_size=0.2, random_state=42)

# 3. Normalisasi menggunakan MinMaxScaler
scaler_nb = MinMaxScaler()
X_train_nb_scaled = scaler_nb.fit_transform(X_train_nb)
X_test_nb_scaled = scaler_nb.transform(X_test_nb)

# 4. Hyperparameter Tuning dengan rentang var_smoothing yang lebih luas (75 variasi)
param_grid_nb = {
    'var_smoothing': np.logspace(-11, -2, num=75)
}

print("--- Memulai Hyperparameter Tuning Naïve Bayes ---")
grid_search_nb = GridSearchCV(GaussianNB(), param_grid_nb, cv=5, scoring='accuracy', n_jobs=-1)
grid_search_nb.fit(X_train_nb_scaled, y_train_nb)

# 5. Prediksi & Evaluasi
best_nb = grid_search_nb.best_estimator_
nb_pred = best_nb.predict(X_test_nb_scaled)

print(f"Parameter Terbaik Naïve Bayes: {grid_search_nb.best_params_}")
print(f"Akurasi Naïve Bayes (Tuned): {accuracy_score(y_test_nb, nb_pred) * 100:.2f}%")
print("Confusion Matrix:\n", confusion_matrix(y_test_nb, nb_pred))
```
