```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

gender = ['perempuan', 'laki-laki', 'laki-laki', 'perempuan', 'perempuan']
df = pd.DataFrame({'gender': gender})

encoder = OneHotEncoder()
encoded_df = pd.DataFrame(
    encoder.fit_transform(df[['gender']]).toarray(),
    columns=encoder.get_feature_names_out(['gender'])
)

df_encoded = pd.concat([df, encoded_df], axis=1)
print(df_encoded)
```

```python
import io
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier

# Membuat data iklan_sosmed.csv langsung di dalam kode
csv_sosmed = """User ID,Gender,Age,EstimatedSalary,Purchased
15624510,Male,19,19000,0
15810944,Male,35,20000,0
15668575,Female,26,43000,0
15603246,Female,27,57000,0
15804002,Male,19,76000,0
15728773,Male,27,58000,0
15598044,Female,27,84000,0
15694829,Female,32,150000,1
15600575,Male,25,33000,0
15727311,Female,35,65000,0
15570769,Female,26,80000,0
15606274,Female,26,52000,0
15746139,Male,20,86000,0
15704983,Male,32,18000,0
15628972,Male,18,82000,0
"""

# Load Dataset
df = pd.read_csv(io.StringIO(csv_sosmed))
print("--- Dataset Awal ---")
print(df.head())

# Encoding Data Kategorik (Gender)
encoder = OneHotEncoder(drop='first', sparse_output=False)
gender_encoded = encoder.fit_transform(df[['Gender']])
df_encoded = pd.DataFrame(gender_encoded, columns=encoder.get_feature_names_out(['Gender']))

# Penggabungan dataframe
df_clean = pd.concat([df.drop(columns=['Gender']), df_encoded], axis=1)

# Handling Outliers
z_scores = np.abs((df_clean['EstimatedSalary'] - df_clean['EstimatedSalary'].mean()) / df_clean['EstimatedSalary'].std())
df_clean = df_clean[z_scores < 3]

# Memisahkan Features (X) dan Target/Label (y)
X = df_clean[['Age', 'EstimatedSalary'] + list(df_encoded.columns)]
y = df_clean['Purchased']

# Normalization / Standardization pada Fitur Numerik
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split Data (Training & Testing)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print(f"\nJumlah Data Train: {X_train.shape[0]}")
print(f"Jumlah Data Test: {X_test.shape[0]}")

# Evaluasi Model dengan Cross Validation
dt_model = DecisionTreeClassifier(random_state=42)
scores = cross_val_score(dt_model, X_scaled, y, cv=2)

print("\n--- Hasil Evaluasi Cross Validation ---")
print("Skor per Fold:", scores)
print("Rata-rata Akurasi: {:.2f}%".format(scores.mean() * 100))
```
